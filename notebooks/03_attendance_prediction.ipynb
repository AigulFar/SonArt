# 03. Прогноз посещаемости и подбор организации

```python
import sys
from pathlib import Path
sys.path.append(str(Path("../AI_section").resolve()))
from ai_predict2 import load_data,train_event_success_model,recommend_organizations
events,institutions,attendance,venues=load_data()
model,mae=train_event_success_model(events,institutions,attendance)
print(f"MAE: {mae:.2f} процентных пункта")
```

```python
recommendations=recommend_organizations(
    model=model,
    event_type="Лекция",
    date="2026-10-10",
    start_time="18:00",
    duration_min=90,
    price_rub=300,
    target_age_min=14,
    target_age_max=30,
    institutions=institutions,
    venues=venues,
    top_n=5,
)
display(recommendations)
```
