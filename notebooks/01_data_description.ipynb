## Связи

- `institutions_official.institution_id` → `events_synthetic.institution_id`
- `institutions_official.institution_id` → `venues_synthetic.institution_id`
- `events_synthetic.event_id` → `attendance_synthetic.event_id`
- `events_synthetic.event_id` → `visitor_events_synthetic.event_id`
- `events_synthetic.event_id` → `feedback_synthetic.event_id`
- `visitor_events_synthetic.visitor_id` → `audience_synthetic.visitor_id`
- `venues_synthetic.venue_id` → `events_synthetic.venue_id`

```python
institutions=pd.read_excel(DATA_PATH,"institutions_official")
events=pd.read_excel(DATA_PATH,"events_synthetic")
attendance=pd.read_excel(DATA_PATH,"attendance_synthetic")
venues=pd.read_excel(DATA_PATH,"venues_synthetic")
ve=pd.read_excel(DATA_PATH,"visitor_events_synthetic")
feedback=pd.read_excel(DATA_PATH,"feedback_synthetic")

assert set(events.institution_id)<=set(institutions.institution_id)
assert set(events.venue_id)<=set(venues.venue_id)
assert set(attendance.event_id)<=set(events.event_id)
assert set(ve.event_id)<=set(events.event_id)
assert set(feedback.event_id)<=set(events.event_id)
assert set(feedback.visitor_id)<=set(ve.visitor_id)

print("OK: внешние ключи корректны.")
print("Уникальных посетителей:",ve.visitor_id.nunique())
print("Посещений:",len(ve))
```
