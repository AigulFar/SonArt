# 05. ML → рекомендация организации → LLM

Сначала получаем числовой результат ML. Затем этот результат можно передать в существующий `ai_rec.py` для формирования текста.

```python
import sys, json
from pathlib import Path
sys.path.append(str(Path('../AI_section').resolve()))
from ai_pipeline import build_ml_context
```

```python
context = build_ml_context(
    event_type='Лекция',
    date='2026-10-10',
    start_time='18:00',
    duration_min=90,
    price_rub=300,
    target_age_min=14,
    target_age_max=30,
    top_n=5,
)
print(json.dumps(context, ensure_ascii=False, indent=2))
```

## Что передаём LLM

В `ai_rec.py` можно передать данные из `context['recommendations']` вместе с описанием мероприятия. LLM должна объяснить выбор организации и формата, но не менять числа, которые посчитала ML-модель.
